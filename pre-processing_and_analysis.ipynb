```python
# remove useless row
df_V1 = pd.read_csv("data_V1.csv", skiprows=[1])
# remove header row
df_V1 = df_V1[~df_V1['ID'].astype(str).str.contains('ImportId', na=False)].copy()


# Arabic  vs non-arabic
df_V1['is_arabic_speaker'] = df_V1['arabic_speaker'].str.lower().str.strip() == 'yes'
df_V1.head()
```

```python
# Correcting Wrong ID number
df_V1 = df_V1[df_V1['ID'] != 'try'] # drop ID try
df_V1['ID'] = df_V1['ID'].replace('3049865d', '04') # change this ID

df_V1.loc[(df_V1['ID'] == '08') & (df_V1['gender'] == 'Female'), 'ID'] = '09' # swap wrong ID 08 - 09
df_V1
```

```python
df_V1['ID'].value_counts().sort_index() # just check
```

```python
# define all rectangle columns
rect_columns = [
    'rectangleData_item1_english_messy',
    'rectangleData_item1_english_clear',
    'rectangleData_item2_english_messy',
    'rectangleData_item2_english_clear',
    'rectangleData_item3_arabic_messy',
    'rectangleData_item3_arabic_clear',
    'rectangleData_item4_arabic_messy',
    'rectangleData_item4_arabic_clear',
]

# # Map columns to metadata
column_metadata = {
    'rectangleData_item1_english_messy': {'item': 'Item1', 'language': 'English', 'condition': 'Messy'},
    'rectangleData_item1_english_clear': {'item': 'Item1', 'language': 'English', 'condition': 'Clear'},
    'rectangleData_item2_english_messy': {'item': 'Item2', 'language': 'English', 'condition': 'Messy'},
    'rectangleData_item2_english_clear': {'item': 'Item2', 'language': 'English', 'condition': 'Clear'},
    'rectangleData_item3_arabic_messy': {'item': 'Item3', 'language': 'Arabic', 'condition': 'Messy'},
    'rectangleData_item3_arabic_clear': {'item': 'Item3', 'language': 'Arabic', 'condition': 'Clear'},
    'rectangleData_item4_arabic_messy': {'item': 'Item4', 'language': 'Arabic', 'condition': 'Messy'},
    'rectangleData_item4_arabic_clear': {'item': 'Item4', 'language': 'Arabic', 'condition': 'Clear'},
}
```

```python
# Parse all rectangles
all_rectangles = []

for idx, row in df_V1.iterrows():
    response_id = row['ID']
    is_arabic = row['is_arabic_speaker']

    for col in rect_columns:
        if pd.notna(row[col]):
            try:
                # Parse JSON
                data = json.loads(row[col])

                # Get metadata
                meta = column_metadata[col]

                # Extract each rectangle
                for rect_idx, rect in enumerate(data.get('rects', []), 1):
                    all_rectangles.append({
                        'ID': response_id,
                        'is_arabic_speaker': is_arabic,
                        'speaker_group': 'Arabic Speaker' if is_arabic else 'Non-Arabic Speaker',
                        'Item': meta['item'],
                        'Language': meta['language'],
                        'Condition': meta['condition'],
                        'img_width': data.get('imgWidth'),
                        'img_height': data.get('imgHeight'),
                        'total_rectangles': data.get('rectangleCount'),
                        'rectangle_number': rect_idx,
                        'rect_x': rect['x'],
                        'rect_y': rect['y'],
                        'rect_width': rect['width'],
                        'rect_height': rect['height'],
                        'rect_area': rect['width'] * rect['height'],
                        'rect_center_x': rect['x'] + rect['width']/2,
                        'rect_center_y': rect['y'] + rect['height']/2,
                    })
            except (json.JSONDecodeError, KeyError, TypeError) as e:
                print(f"  Warning: Could not parse {col} for {response_id}: {e}")

# Create DataFrame
df_rectangles = pd.DataFrame(all_rectangles)
df_rectangles
```

```python
# Calculate additional metrics
df_rectangles['rect_percent_of_image'] = (
    df_rectangles['rect_area'] /
    (df_rectangles['img_width'] * df_rectangles['img_height']) * 100
)

df_rectangles['aspect_ratio'] = df_rectangles['rect_width'] / df_rectangles['rect_height']

# Quadrant classification
df_rectangles['quadrant_x'] = (df_rectangles['rect_center_x'] > df_rectangles['img_width']/2).map({
    True: 'Right', False: 'Left'
})
df_rectangles['quadrant_y'] = (df_rectangles['rect_center_y'] > df_rectangles['img_height']/2).map({
    True: 'Bottom', False: 'Top'
})
df_rectangles['quadrant'] = df_rectangles['quadrant_y'] + '-' + df_rectangles['quadrant_x']

df_rectangles
```

In pandas' .agg() method, 'first' is an aggregation function that returns the first non-null value for each group within the specified column. It's useful when you have group-specific metadata that is constant for all rows within that group, and you just need to pick one representative value (the first one encountered).



```python
# Aggregate by participant, item
final_df = df_combined_rectangles.groupby(
    ['ID', 'Item', 'Language', 'Condition', 'is_arabic_speaker']
).agg({
    'total_rectangles': 'first',
    'rect_area': ['mean', 'median', 'std', 'sum'],
    'rect_width': 'mean',
    'rect_height': 'mean',
    'aspect_ratio': 'mean',
    'rect_percent_of_image': 'sum',  # Total coverage
    'img_width': 'first',
    'img_height': 'first',
}).reset_index()

final_df.head()
```

```python
# Flatten only the aggregated column names (not the groupby columns)
new_cols = []
for col in final_df.columns:
    if isinstance(col, tuple):
        # Flatten tuple columns from aggregation
        new_cols.append('_'.join(str(c) for c in col).strip('_'))
    else:
        # Keep simple column names as-is
        new_cols.append(col)
final_df.columns = new_cols

final_df
```

```python
# Add speaker_group column
final_df['speaker_group'] = final_df['is_arabic_speaker'].map({
    True: 'Arabic Speaker', False: 'Non-Arabic Speaker'
})

len(final_df['ID'])
final_df.head()
```

```python
# Flatten MultiIndex columns in participant_summary before merging
# This ensures both DataFrames have single-level columns for a successful merge.
if isinstance(final_df.columns, pd.MultiIndex):
    new_cols = []
    for col in final_df.columns:
        if isinstance(col, tuple):
            new_cols.append('_'.join(str(c) for c in col).strip('_'))
        else:
            new_cols.append(col)
    final_df.columns = new_cols

merged_data = pd.merge(final_df, demographics, on='ID', how='left')
# Dropping 'arabic_speaker' from the merged data as it's likely redundant with 'is_arabic_speaker' and the previous attempt to drop it failed due to the merge error.
merged_data = merged_data.drop(columns=['arabic_speaker'])
merged_data.head()
```

### Rectangle Density Heatmap

To visualize the areas where rectangles are most frequently drawn, a heatmap can be generated. This heatmap shows the density of rectangle centers across the canvas.

```python
import numpy as np

# Get image dimensions (assuming they are constant as observed from previous data)
img_width = df_combined_rectangles['img_width'].iloc[0]
img_height = df_combined_rectangles['img_height'].iloc[0]

# Define the number of bins for the heatmap grid
n_bins_x = 50
n_bins_y = 50

# Create a 2D histogram of rectangle centers
heatmap, xedges, yedges = np.histogram2d(
    df_combined_rectangles['rect_center_x'],
    df_combined_rectangles['rect_center_y'],
    bins=[n_bins_x, n_bins_y],
    range=[[0, img_width], [0, img_height]]
)

# Transpose the heatmap to align with image coordinates (origin top-left)
heatmap = heatmap.T

plt.figure(figsize=(12, 9))
sns.heatmap(heatmap, cmap='hot_r', cbar_kws={'label': 'Rectangle Density'})

plt.title('Heatmap of Rectangle Density')
plt.xlabel('X-coordinate Bins')
plt.ylabel('Y-coordinate Bins')

# Adjust tick labels to represent actual coordinates (approximate)
x_labels = [f'{int(l)}' for l in np.linspace(0, img_width, n_bins_x + 1)[:-1]]
y_labels = [f'{int(l)}' for l in np.linspace(0, img_height, n_bins_y + 1)[:-1]]

# Set tick locations to match the number of labels (every 5th bin)
plt.xticks(np.arange(0, n_bins_x, 5) + 0.5, x_labels[::5], rotation=45)
plt.yticks(np.arange(0, n_bins_y, 5) + 0.5, y_labels[::5], rotation=0)

plt.gca().invert_yaxis() # Invert y-axis to match image coordinates (origin top-left)
plt.tight_layout()
plt.show()
```
